# 11주차 후보 목록 검토
문헌으로 분류한 30종의 그룹별 표본 수와 시험 수집 후보를 표시합니다.

```python
from pathlib import Path
import pandas as pd

# 노트북을 프로젝트 루트 또는 11주차 폴더에서 실행합니다.
project_dir = Path.cwd()
if project_dir.name == 'week11':
    project_dir = project_dir.parent
species = pd.read_csv(project_dir / 'data/processed/species_metadata.csv')
group_counts = pd.crosstab(species['taxonomic_group'], species['photosynthesis_type'])
group_counts.reindex(['Monocot', 'Eudicot'])
```

```python
pilot_species = ['Oryza sativa', 'Flaveria pringlei', 'Zea mays', 'Flaveria bidentis']
columns = ['scientific_name', 'family', 'taxonomic_group', 'photosynthesis_type']
species.loc[species['scientific_name'].isin(pilot_species), columns]
```

```python
# 사전 지정한 속 내 비교와 같은 과의 Cleomaceae 대비를 표시합니다.
genera = species['scientific_name'].str.split().str[0]
matches = genera.isin(['Flaveria', 'Atriplex', 'Cyperus']) | species['family'].eq('Cleomaceae')
species.loc[matches, columns].sort_values(['family', 'photosynthesis_type', 'scientific_name']).reset_index(drop=True)
```
